```python
from pathlib import Path

# Raíz del repo (sirve si el notebook se abre desde notebooks/ o desde la raíz)
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATASET = ROOT / "dataset"
print("Repo:", ROOT)
```

```python
from ultralytics import YOLO

model = YOLO(ROOT / "weights" / "best.pt")
print(model.names)
```

## 1. Imágenes de prueba
Necesita el dataset en `dataset/` (ver Parte 1). `conf=0.7` deja solo las detecciones más seguras.

```python
model.predict(
    source=str(DATASET / "test" / "images"),
    imgsz=640,
    conf=0.7,
    save=True,
    project=str(ROOT / "predicciones"),
    name="test",
)
```

## 2. Video
Pon la ruta de tu video. `stream=True` procesa cuadro por cuadro sin llenar la memoria; el video con las cajas queda en `predicciones/video/`.

```python
video = ROOT / "videos" / "mi_video.mp4"   # cambia por tu archivo

con_deteccion = 0
for r in model.predict(source=str(video), imgsz=640, conf=0.6, stream=True, save=True,
                       project=str(ROOT / "predicciones"), name="video", verbose=False):
    con_deteccion += len(r.boxes) > 0
print("Cuadros con al menos una entrada detectada:", con_deteccion)
```
